```python
from pathlib import Path
import json

from peft import PeftModel

from src.model import MODEL_NAME, load_qwen_model
from src.train_lora import (
    split_tiser_train_validation,
    evaluate_validation_dataset,
)

checkpoint = "/content/drive/MyDrive/TISER/checkpoints/train_lora/checkpoint-49040"

bundle = load_qwen_model(MODEL_NAME, device_map="auto")
tokenizer = bundle.tokenizer

model = PeftModel.from_pretrained(
    bundle.model,
    checkpoint,
)

_, validation_dataset = split_tiser_train_validation(
    validation_ratio=0.1,
    seed=42,
)

metrics = evaluate_validation_dataset(
    model=model,
    tokenizer=tokenizer,
    validation_dataset=validation_dataset,
    batch_size=16,
    max_new_tokens=2048,
    epoch=2,
)

output_path = Path(
    "/content/drive/MyDrive/TISER/checkpoints/train_lora/epoch2_validation_metrics.json"
)

with output_path.open("w", encoding="utf-8") as f:
    json.dump(metrics, f, indent=2)

print(json.dumps(metrics, indent=2))
print(f"Saved to: {output_path}")
```
